```python
student_data['high_score'] = student_data['mark'] >= 75
print(student_data)
```

```python
student_data.rename(
    columns={'name': 'student_name'},
    inplace=True
)
print(student_data)
```

```python
student_data.drop(columns=['high_score'], inplace=True, errors='ignore')
print(student_data)
```

```python
student_data.groupby('class')['mark'].mean()
```

```python
student_data.groupby('gender').size()
```

```python
student_data.groupby('class')['mark'].max()
```

```python
pd.pivot_table(
    student_data,
    values='mark',
    index='class',
    columns='gender',
    aggfunc='mean'
)
```

```python
student_data['performance'] = pd.cut(
    student_data['mark'],
    bins=[-float('inf'), 60, 70, 80, float('inf')],
    labels=[
        'Needs Improvement',
        'Satisfactory',
        'Good',
        'Excellent'
    ],
    right=False
)
print(student_data)
```

```python
student_data = student_data.sort_values(
    by='mark',
    ascending=True
)
print(student_data)
```

```python
student_data.to_csv(
    'student_performance_report.csv',
    index=False
)
```
